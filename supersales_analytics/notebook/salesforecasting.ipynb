```python
# Create time-based features

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
df["Year-Month"] = df["Order Date"].dt.to_period("M")

# Shipping duration
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# Profit margin
df["Profit Margin"] = (
    df["Profit"] / df["Sales"]
) * 100

print("Feature engineering completed!")
```

```python
# Create monthly sales time series

forecast_df = (
    df.groupby("Year-Month")["Sales"]
      .sum()
      .reset_index()
)

# Convert Period to datetime
forecast_df["ds"] = forecast_df["Year-Month"].dt.to_timestamp()

# Prophet expects the target column to be called "y"
forecast_df["y"] = forecast_df["Sales"]

# Keep only Prophet columns
forecast_df = forecast_df[["ds", "y"]]

# Sort chronologically
forecast_df = forecast_df.sort_values("ds").reset_index(drop=True)

forecast_df.head()
```

```python
print("Number of observations:", len(forecast_df))
print("Start date:", forecast_df["ds"].min())
print("End date:", forecast_df["ds"].max())
print("Sorted:", forecast_df["ds"].is_monotonic_increasing)
```

```python
# Aggregate sales by day

daily_sales = (
    df.groupby("Order Date")["Sales"]
      .sum()
      .reset_index()
)

daily_sales.columns = ["ds", "y"]

daily_sales.head()
```

```python
print("Total calendar days:",
      (df["Order Date"].max() - df["Order Date"].min()).days + 1)

print("Days with sales:", len(daily_sales))
```

```python
# Create complete daily date range

complete_dates = pd.date_range(
    start=daily_sales["ds"].min(),
    end=daily_sales["ds"].max(),
    freq="D"
)

# Reindex using all calendar dates
daily_sales = (
    daily_sales
    .set_index("ds")
    .reindex(complete_dates, fill_value=0)
    .rename_axis("ds")
    .reset_index()
)

daily_sales.head()
```

```python
test_days = 90

train = daily_sales.iloc[:-test_days].copy()
test = daily_sales.iloc[-test_days:].copy()

print("Training data:")
print(train["ds"].min(), "→", train["ds"].max())

print("\nTesting data:")
print(test["ds"].min(), "→", test["ds"].max())

print("\nTrain rows:", len(train))
print("Test rows:", len(test))
```

```python
model = Prophet()

model.fit(train)

print("Prophet model trained successfully!")
```

```python
future = model.make_future_dataframe(
    periods=test_days,
    freq="D"
)

forecast = model.predict(future)
```

```python
forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail(90)
```

```python
fig1 = model.plot(forecast)

plt.title("Sales Forecast using Prophet")
plt.xlabel("Date")
plt.ylabel("Sales ($)")
plt.show()
```

```python
# Get Prophet predictions for test dates

predictions = forecast[
    forecast["ds"].isin(test["ds"])
][["ds", "yhat"]].copy()

# Merge actual and predicted values
results = test.merge(predictions, on="ds")

results.head()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(results["y"], results["yhat"])

rmse = np.sqrt(
    mean_squared_error(results["y"], results["yhat"])
)

# Avoid division by zero because we have zero-sales days
non_zero = results["y"] != 0

mape = np.mean(
    np.abs(
        (results.loc[non_zero, "y"] -
         results.loc[non_zero, "yhat"])
        / results.loc[non_zero, "y"]
    )
) * 100

print(f"MAE  : ${mae:,.2f}")
print(f"RMSE : ${rmse:,.2f}")
print(f"MAPE : {mape:.2f}%")
```

```python
fig = px.line(
    results,
    x="ds",
    y=["y", "yhat"],
    title="Actual vs Predicted Sales"
)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Sales ($)",
    hovermode="x unified"
)

fig.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# ---- FINAL MODEL: Monthly Sales Forecast ----

# 1. Aggregate daily sales into monthly totals
monthly_data = daily_sales.set_index('ds').resample('M')['y'].sum().reset_index()
monthly_data.columns = ['ds', 'y']

# 2. Split: hold out last 6 months for testing
monthly_train = monthly_data.iloc[:-6]
monthly_test = monthly_data.iloc[-6:]

# 3. Train Prophet
monthly_model = Prophet(yearly_seasonality=True, weekly_seasonality=False, daily_seasonality=False)
monthly_model.fit(monthly_train)

# 4. Forecast forward (covers test months + 6 genuinely new future months)
monthly_future = monthly_model.make_future_dataframe(periods=12, freq='M')
monthly_forecast = monthly_model.predict(monthly_future)

# 5. Evaluate on test months
monthly_forecast_test = monthly_forecast.set_index('ds').loc[monthly_test['ds']]
actual_values = monthly_test.set_index('ds')['y']
predicted_values = monthly_forecast_test['yhat']

mae = mean_absolute_error(actual_values, predicted_values)
rmse = np.sqrt(mean_squared_error(actual_values, predicted_values))
r2 = r2_score(actual_values, predicted_values)
mape = (abs((actual_values - predicted_values) / actual_values)).mean() * 100
accuracy = 100 - mape

print("========== FINAL MODEL: MONTHLY SALES FORECAST ==========")
print(f"MAE:      {mae:,.2f}")
print(f"RMSE:     {rmse:,.2f}")
print(f"R²:       {r2:.3f}")
print(f"MAPE:     {mape:.1f}%")
print(f"Accuracy: {accuracy:.1f}%")
print("===========================================================")
```

```python
# ============================================================
# EXPORT FINAL MONTHLY PROPHET FORECAST FOR TABLEAU
# ============================================================

# Historical monthly sales
historical = monthly_data.copy()

# Prophet forecast
forecast_tableau = monthly_forecast.copy()

# Keep only the useful Prophet columns
forecast_tableau = forecast_tableau[
    ["ds", "yhat", "yhat_lower", "yhat_upper"]
].copy()

# Rename columns
forecast_tableau = forecast_tableau.rename(
    columns={
        "ds": "Date",
        "yhat": "Forecast Sales",
        "yhat_lower": "Forecast Lower",
        "yhat_upper": "Forecast Upper"
    }
)

# Prepare historical data
historical = historical.rename(
    columns={
        "ds": "Date",
        "y": "Actual Sales"
    }
)

# Keep only required columns
historical = historical[["Date", "Actual Sales"]]

# Combine historical sales with forecast
prophet_tableau = forecast_tableau.merge(
    historical,
    on="Date",
    how="left"
)

# Identify Actual vs Forecast
last_actual_date = historical["Date"].max()

prophet_tableau["Data Type"] = np.where(
    prophet_tableau["Date"] <= last_actual_date,
    "Actual",
    "Forecast"
)

# Add date fields for Tableau
prophet_tableau["Year"] = prophet_tableau["Date"].dt.year
prophet_tableau["Month"] = prophet_tableau["Date"].dt.month
prophet_tableau["Month Name"] = prophet_tableau["Date"].dt.month_name()
prophet_tableau["Quarter"] = (
    "Q" + prophet_tableau["Date"].dt.quarter.astype(str)
)

# Sort by date
prophet_tableau = prophet_tableau.sort_values("Date")

# Save
prophet_tableau.to_csv(
    "prophet_forecast_tableau.csv",
    index=False
)

print("========================================")
print("✅ PROPHET TABLEAU FILE CREATED")
print("========================================")

print("\nRows:", len(prophet_tableau))
print("\nColumns:")
print(prophet_tableau.columns.tolist())

print("\nLast 10 rows:")
print(prophet_tableau.tail(10))

print("\nData Type:")
print(prophet_tableau["Data Type"].value_counts())

print("\nModel Accuracy:", round(accuracy, 2), "%")
print("MAPE:", round(mape, 2), "%")

# Download to your computer
from google.colab import files
files.download("prophet_forecast_tableau.csv")
```
